### FUNCION DE ORDEN SUPERIOR MAP

```python
numeros = ['uno','dos','tres','cuatro']
```

```python
NUMEROS = map (str.upper, numeros)
```

```python
NUMEROS
```

```python
for numero in NUMEROS:
    print(numero)
```

EXPRESION GENERADORA

expresion for elemento in elementos if condicion

```python
NUMEROS2=(numero.upper() for numero in numeros)
```

```python
for numero in NUMEROS2:
    print(numero)
```

```python
NUMEROS2=(numero.upper() for numero in numeros if len(numero) > 3)
```

```python
for numero in NUMEROS2:
    print(numero)
```

filter(funcion, iterable) es una funcion de orden superior que devuelve los elementos de iterable para los cuales la funcion evalua a True

```python
cortos=filter(lambda numero: len(numero)< 4, numeros)
```

```python
for corto in cortos:
    print(corto)
```

```python
cortos=(elemento.title() for elemento in numeros if len(elemento)< 4)
```

```python
for corto in cortos:
    print(corto)
```

```python
from functools import reduce
```

```python
help(reduce)
```

```python
def factorial(N):
    return reduce(lambda x,y: x*y,range(2,N+1))
```

```python
for N in range (2,10):
    print(f"{N}! = {factorial(N)}")
```

```python
class Alumno: # tiene que ser self el primero
    def __init__(self, nombre, apellido, id):
        self.nombre, self.apellido, self.id = nombre, apellido, id
    def __str__(self):
        return f"{self.id}: {self.nombre} {self.apellido}"
    __repr__=__str__
    def __call__(self, saludo):
        print(f'Hola soy {self.nombre} {self.apellido} , {saludo}')
```

```python
alumnos = [Alumno('David','Bueno',3573),Alumno('Eduard','Peñas',6969),Alumno('Pau','Dresaire',7779),Alumno('Oriol','Gimenez',4444)]
```

```python
sorted(alumnos, key=lambda alumno: alumno.apellido)
```

```python
alumnos [0] ('kitipasa') #usamos el metodo call creado en la clase para crear un 'saludo'
```

```python
alumnos [0].__getattribute__('nombre')
```

```python
def ordena_atributo(iterable, clave):
    return sorted(iterable, key =lambda elemento: elemento.__getattribute__(clave))
```

```python
ordena_atributo(alumnos, 'id')
```

### DECORADOR

Toma como entrada una fincoin y devuelve otra. Devuelva la orginal 'decorada', con un emboltorio

```python
def decora(funcion):
    def envoltorio(*args, **dicc):
        print('Ejecuto', funcion)
        return funcion(*args, **dicc)
    return envoltorio
```

```python
decorada = decora(print)
```

```python
print(1,2,3, sep='\t')
```

```python
decorada(1, 2, 3, sep='\t')
```

```python
@decora
def ordena_atributo(iterable, clave):
    return sorted(iterable, key =lambda elemento: elemento.__getattribute__(clave))
```

```python
ordena_atributo(alumnos, 'nombre')
```
