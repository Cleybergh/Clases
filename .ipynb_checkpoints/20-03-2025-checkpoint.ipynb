```python
def function(primer, *tupla):
    print(primer)
    print(tupla)
    for element in tupla:
        print(element)
```

```python
function()
```

```python
function(1,2,3,4,5)
```

```python
def function(sep='->',**cosas):
    print (cosas)
    for cosa in cosas:
        print(f'{cosa} {sep} {cosas[cosa]}')
```

```python
function(seat='blanco',fiat='azul', ford='negro') 
```

```python
function(sep = ':',seat='blanco',fiat='azul', ford='negro') 
```
